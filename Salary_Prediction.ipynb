```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
```

```python
data["Gender"] = label_encoder.fit_transform(data["Gender"])
```

```python
data["Education Level"] = label_encoder.fit_transform(data["Education Level"])
```

```python
data["Job Title"] = label_encoder.fit_transform(data["Job Title"])
```

```python
x1=data["Age"].mean()
x1
```

```python
x2=data["Gender"].mode()
x2
```

```python
x3=data["Education Level"].mode()
x3
```

```python
x4=data["Job Title"].mode()
x4
```

```python
x5=data["Experience"].mode()
x5
```

```python
x6=data["Salary"].mode()
x6
```

```python
x7=data["Name"].mode()
x7
```

```python
data["Age"]=data["Age"].fillna(x1, inplace = True)
data["Gender"]=data["Gender"].fillna(x2, inplace = True)
data["Education Level"]=data["Education Level"].fillna(x3, inplace = True)
data["Job Title"]=data["Job Title"].fillna(x4, inplace = True)
data["Experience"]=data["Experience"].fillna(x5, inplace = True)
data["Salary"]=data["Salary"].fillna(x6, inplace = True)
data["Name"]=data["Name"].fillna(x7, inplace = True)
```

```python
missingvalue = data.isnull().sum()
missingvalue
```
